```python
import pandas as pd
#定义Series有多种方式
s1 = pd.Series([10, 20, 30, 40, 50], index = ["a", "b", "c", "d", "e"])
s2 = pd.Series({"a":10, "b":20, "c":30, "d":40, "e":50})
#Series也能使用Index, Values, size, dtype, shape这种基本属性
print(s1.index)
print(s1.values)
print(s1.size)
print(s1.dtype)
print(s1.shape)
```

```python
#读取数据
df1 = pd.read_csv("user.csv")
print(df1)
print("\n")
#使用usecols可以选择只使用哪些列
df2 = pd.read_csv("product_sales.csv")
print(df2)
print("\n")
df3 = pd.read_csv("product_sales.csv", usecols=["商品名称", "单价", "销售量"])
print(df3)
print("\n")
#数据处理
df3["销售金额"] = df3["单价"] * df3["销售量"]
print(df3)
print("\n")
#导出数据为csv
df3.to_csv("sales_result.csv", index=False) #index=Fales指的是不写入索引列
```
