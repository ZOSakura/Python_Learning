```python
import pandas as pd
#创建DataFrame对象
df = pd.DataFrame([
        {"姓名": "张三", "语文": 90, "数学": 88, "英语": 78},
        {"姓名": "李四", "语文": 67, "数学": 78, "英语": 91},
        {"姓名": "王五", "语文": 91, "数学": 66, "英语": 77},
        {"姓名": "海猫", "语文": 100, "数学": 67, "英语": 100}
])
print(type(df["姓名"])) #DataFrame拿出来的单独的一列的类型是Series
print(f"语文最高分: {df["语文"].max()}, 数学最高分: {df["数学"].max()}, 英语最高分: {df["英语"].max()}")
print(f"数学平均分: {df["数学"].mean():.2f}") # max最大值 min最小值 mean平均值
print(df) #单成一列的数字是Index，即索引
#DataFrame支持多种定义方式
df1 = pd.DataFrame({
    "姓名":["张三", "李四", "王五", "海猫"],
    "语文":[90, 67, 91, 100],
    "数学":[88, 78, 66, 67],
    "英语":[78, 91, 77, 100]
})
print(df1)
df2 = pd.DataFrame([
    ["张三", 90, 88, 78],
    ["李四", 67, 78, 91],
    ["王五", 91, 66, 77],
    ["海猫", 100, 67, 100]
], columns=["姓名", "语文", "数学", "英语"], index=["A", "B", "C", "D"]) #Index也支持自定义
print(df2)
#获取Index
print(df.index) #通过.tolist可以封装成list
#获取colums(表头)
print(df.columns)
#获取vaule(值)
print(df.values)
#获取每一列的类型
print(df.dtypes)
#获取单元格的数量
print(df.size)
#获取数据维度(行, 列)
print(df.shape)
```
